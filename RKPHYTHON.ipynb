```python
units_str = input("Enter the number of units consumed: ")
customer_type = input("Enter customer type (e.g., 'residential' or 'commercial'): ")

units = int(units_str)
total_cost = 0.0

print(f"\nCalculating price for {units} units and customer type '{customer_type}'...")

# Residential customer pricing
if customer_type.lower() == 'residential':
    if units <= 100:
        total_cost = units * 0.10
    elif units <= 300: # units > 100 and <= 300
        total_cost = (100 * 0.10) + ((units - 100) * 0.15)
    else: # units > 300
        total_cost = (100 * 0.10) + (200 * 0.15) + ((units - 300) * 0.20)
    print(f"Residential Customer Total Cost: ${total_cost:.2f}")

# Commercial customer pricing
elif customer_type.lower() == 'commercial':
    if units <= 500:
        total_cost = units * 0.12
    else: # units > 500
        total_cost = (500 * 0.12) + ((units - 500) * 0.18)
    print(f"Commercial Customer Total Cost: ${total_cost:.2f}")

else:
    print("Invalid customer type entered. Please enter 'residential' or 'commercial'.")
```

PRINT NUMBER TABLE


```python
num_str = input("Enter a number to see its multiplication table: ")
num = int(num_str)

print(f"\nMultiplication Table for {num}:")
for i in range(1, 11): # Loop from 1 to 10
    print(f"{num} x {i} = {num * i}")
```

```python
num_str_while = input("Enter a number for its multiplication table (using while loop): ")
num_while = int(num_str_while)

print(f"\nMultiplication Table for {num_while} (using while loop):")

i = 1
while i <= 10:
    print(f"{num_while} x {i} = {num_while * i}")
    i += 1
```
